# Global Superstore Data Visualization 

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set style for better-looking plots
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Load the data
df = pd.read_csv('Global_Superstore.csv', encoding='latin-1')
print(f"Dataset shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
```

## 1. Basic Data Overview

```python
# Display first few rows
print("First 5 rows of data:")
df.head()
```

```python
# Data types and missing values
print("Data Info:")
df.info()
print("\nMissing Values:")
df.isnull().sum()
```

```python
# Statistical summary
print("Statistical Summary:")
df.describe()
```

## 2. Distribution Plots
### Understanding how values are spread

```python
# Sales Distribution - Histogram
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(df['Sales'], bins=50, color='skyblue', edgecolor='black')
plt.title('Sales Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(df['Profit'], bins=50, color='lightcoral', edgecolor='black')
plt.title('Profit Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Profit Amount')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Box Plot - Identifying outliers
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].boxplot(df['Sales'])
axes[0].set_title('Sales Box Plot', fontweight='bold')
axes[0].set_ylabel('Sales')

axes[1].boxplot(df['Profit'])
axes[1].set_title('Profit Box Plot', fontweight='bold')
axes[1].set_ylabel('Profit')

axes[2].boxplot(df['Quantity'])
axes[2].set_title('Quantity Box Plot', fontweight='bold')
axes[2].set_ylabel('Quantity')

plt.tight_layout()
plt.show()
```

## 3. Categorical Data Analysis
### Exploring categories and segments

```python
# Count Plot - Category Distribution
plt.figure(figsize=(10, 6))
category_counts = df['Category'].value_counts()
plt.bar(category_counts.index, category_counts.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1'])
plt.title('Number of Orders by Category', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=0)
for i, v in enumerate(category_counts.values):
    plt.text(i, v + 100, str(v), ha='center', fontweight='bold')
plt.show()
```

```python
# Pie Chart - Market Distribution
plt.figure(figsize=(10, 8))
market_counts = df['Market'].value_counts()
colors = plt.cm.Set3(range(len(market_counts)))
plt.pie(market_counts.values, labels=market_counts.index, autopct='%1.1f%%', 
        startangle=90, colors=colors)
plt.title('Market Distribution', fontsize=14, fontweight='bold')
plt.axis('equal')
plt.show()
```

```python
# Segment Analysis
plt.figure(figsize=(10, 6))
segment_counts = df['Segment'].value_counts()
plt.barh(segment_counts.index, segment_counts.values, color=['#95E1D3', '#F38181', '#EAFFD0'])
plt.title('Customer Segment Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Count')
plt.ylabel('Segment')
for i, v in enumerate(segment_counts.values):
    plt.text(v + 100, i, str(v), va='center', fontweight='bold')
plt.show()
```

## 4. Sales Analysis
### Understanding revenue patterns

```python
# Sales by Category
plt.figure(figsize=(12, 6))
sales_by_category = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
plt.bar(sales_by_category.index, sales_by_category.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1'])
plt.title('Total Sales by Category', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Total Sales ($)')
plt.ticklabel_format(style='plain', axis='y')
for i, v in enumerate(sales_by_category.values):
    plt.text(i, v + 10000, f'${v:,.0f}', ha='center', fontweight='bold')
plt.show()
```

```python
# Top 10 Sub-Categories by Sales
plt.figure(figsize=(12, 6))
top_subcategories = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
plt.barh(range(len(top_subcategories)), top_subcategories.values, color='coral')
plt.yticks(range(len(top_subcategories)), top_subcategories.index)
plt.title('Top 10 Sub-Categories by Sales', fontsize=14, fontweight='bold')
plt.xlabel('Total Sales ($)')
plt.ylabel('Sub-Category')
plt.gca().invert_yaxis()
plt.show()
```

```python
# Sales by Market
plt.figure(figsize=(12, 6))
sales_by_market = df.groupby('Market')['Sales'].sum().sort_values(ascending=False)
colors = plt.cm.viridis(np.linspace(0, 1, len(sales_by_market)))
plt.bar(sales_by_market.index, sales_by_market.values, color=colors)
plt.title('Total Sales by Market', fontsize=14, fontweight='bold')
plt.xlabel('Market')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
for i, v in enumerate(sales_by_market.values):
    plt.text(i, v + 10000, f'${v:,.0f}', ha='center', fontweight='bold', rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Profit Analysis
### Understanding profitability

```python
# Profit by Category
plt.figure(figsize=(12, 6))
profit_by_category = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
colors = ['green' if x > 0 else 'red' for x in profit_by_category.values]
plt.bar(profit_by_category.index, profit_by_category.values, color=colors)
plt.title('Total Profit by Category', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Total Profit ($)')
plt.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
for i, v in enumerate(profit_by_category.values):
    plt.text(i, v + 5000, f'${v:,.0f}', ha='center', fontweight='bold')
plt.show()
```

```python
# Profit Margin by Category
category_stats = df.groupby('Category').agg({'Sales': 'sum', 'Profit': 'sum'})
category_stats['Profit_Margin'] = (category_stats['Profit'] / category_stats['Sales']) * 100

plt.figure(figsize=(10, 6))
plt.bar(category_stats.index, category_stats['Profit_Margin'], color=['#FFB6C1', '#87CEEB', '#98FB98'])
plt.title('Profit Margin by Category (%)', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Profit Margin (%)')
plt.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
for i, v in enumerate(category_stats['Profit_Margin'].values):
    plt.text(i, v + 0.5, f'{v:.1f}%', ha='center', fontweight='bold')
plt.show()
```

## 6. Relationship Analysis
### How variables relate to each other

```python
# Scatter Plot - Sales vs Profit
plt.figure(figsize=(12, 6))
plt.scatter(df['Sales'], df['Profit'], alpha=0.5, c=df['Discount'], cmap='coolwarm')
plt.colorbar(label='Discount')
plt.title('Sales vs Profit (colored by Discount)', fontsize=14, fontweight='bold')
plt.xlabel('Sales ($)')
plt.ylabel('Profit ($)')
plt.axhline(y=0, color='red', linestyle='--', linewidth=1, alpha=0.5)
plt.grid(alpha=0.3)
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10, 8))
numeric_cols = ['Sales', 'Quantity', 'Discount', 'Profit', 'Shipping Cost']
correlation = df[numeric_cols].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, fmt='.2f')
plt.title('Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Quantity vs Sales by Category
plt.figure(figsize=(12, 6))
for category in df['Category'].unique():
    category_data = df[df['Category'] == category]
    plt.scatter(category_data['Quantity'], category_data['Sales'], 
                label=category, alpha=0.6, s=50)
plt.title('Quantity vs Sales by Category', fontsize=14, fontweight='bold')
plt.xlabel('Quantity')
plt.ylabel('Sales ($)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 7. Discount Impact Analysis

```python
# Discount Distribution
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(df['Discount'], bins=20, color='orange', edgecolor='black')
plt.title('Discount Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Discount Rate')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
discount_bins = pd.cut(df['Discount'], bins=[0, 0.1, 0.2, 0.3, 1.0], 
                       labels=['0-10%', '10-20%', '20-30%', '30%+'])
avg_profit = df.groupby(discount_bins)['Profit'].mean()
plt.bar(range(len(avg_profit)), avg_profit.values, color='lightgreen', edgecolor='black')
plt.xticks(range(len(avg_profit)), avg_profit.index)
plt.title('Average Profit by Discount Range', fontsize=14, fontweight='bold')
plt.xlabel('Discount Range')
plt.ylabel('Average Profit ($)')
plt.axhline(y=0, color='red', linestyle='--', linewidth=1)

plt.tight_layout()
plt.show()
```

## 8. Shipping Analysis

```python
# Ship Mode Distribution
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
ship_mode_counts = df['Ship Mode'].value_counts()
plt.pie(ship_mode_counts.values, labels=ship_mode_counts.index, autopct='%1.1f%%', 
        startangle=90, colors=plt.cm.Pastel1(range(len(ship_mode_counts))))
plt.title('Ship Mode Distribution', fontsize=14, fontweight='bold')

plt.subplot(1, 2, 2)
avg_shipping = df.groupby('Ship Mode')['Shipping Cost'].mean().sort_values(ascending=False)
plt.barh(range(len(avg_shipping)), avg_shipping.values, color='steelblue')
plt.yticks(range(len(avg_shipping)), avg_shipping.index)
plt.title('Average Shipping Cost by Mode', fontsize=14, fontweight='bold')
plt.xlabel('Average Shipping Cost ($)')
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

## 9. Geographic Analysis

```python
# Top 10 Countries by Sales
plt.figure(figsize=(12, 6))
top_countries = df.groupby('Country')['Sales'].sum().sort_values(ascending=False).head(10)
plt.barh(range(len(top_countries)), top_countries.values, color='teal')
plt.yticks(range(len(top_countries)), top_countries.index)
plt.title('Top 10 Countries by Sales', fontsize=14, fontweight='bold')
plt.xlabel('Total Sales ($)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Region Performance
region_stats = df.groupby('Region').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Order ID': 'count'
}).sort_values('Sales', ascending=False).head(10)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

axes[0].barh(range(len(region_stats)), region_stats['Sales'].values, color='skyblue')
axes[0].set_yticks(range(len(region_stats)))
axes[0].set_yticklabels(region_stats.index)
axes[0].set_title('Top 10 Regions by Sales', fontweight='bold')
axes[0].set_xlabel('Total Sales ($)')
axes[0].invert_yaxis()

axes[1].barh(range(len(region_stats)), region_stats['Profit'].values, 
             color=['green' if x > 0 else 'red' for x in region_stats['Profit'].values])
axes[1].set_yticks(range(len(region_stats)))
axes[1].set_yticklabels(region_stats.index)
axes[1].set_title('Top 10 Regions by Profit', fontweight='bold')
axes[1].set_xlabel('Total Profit ($)')
axes[1].axvline(x=0, color='black', linestyle='-', linewidth=0.5)
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()
```

## 10. Summary Statistics Dashboard

```python
# Key Metrics Summary
print("=" * 60)
print("GLOBAL SUPERSTORE - KEY METRICS SUMMARY")
print("=" * 60)
print(f"\nTotal Orders: {df['Order ID'].nunique():,}")
print(f"Total Customers: {df['Customer ID'].nunique():,}")
print(f"Total Products: {df['Product ID'].nunique():,}")
print(f"\nTotal Sales: ${df['Sales'].sum():,.2f}")
print(f"Total Profit: ${df['Profit'].sum():,.2f}")
print(f"Average Order Value: ${df['Sales'].mean():,.2f}")
print(f"Average Profit per Order: ${df['Profit'].mean():,.2f}")
print(f"\nOverall Profit Margin: {(df['Profit'].sum() / df['Sales'].sum() * 100):.2f}%")
print(f"Average Discount: {df['Discount'].mean():.2%}")
print(f"Average Quantity per Order: {df['Quantity'].mean():.2f}")
print("\n" + "=" * 60)

# Category Performance
print("\nCATEGORY PERFORMANCE:")
print("-" * 60)
category_summary = df.groupby('Category').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Order ID': 'count'
})
category_summary['Profit_Margin_%'] = (category_summary['Profit'] / category_summary['Sales'] * 100)
print(category_summary.to_string())
print("=" * 60)
```
